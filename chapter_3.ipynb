```python
import pandas as pd
```

```python
credit_card = pd.read_csv('Credit_card.csv')
```

```python
credit_card_label=pd.read_csv('Credit_card_label.csv')
```

```python
credit_card
```

```python
credit_card_label
```

```python
data = pd.merge(credit_card, credit_card_label, on='Ind_ID')
```

```python
data
```

```python
clean_data = data.dropna()
```

```python
feature_data = clean_data[['GENDER', 'Annual_income', 'EDUCATION', 'Marital_status', 'Type_Occupation', 'label']]
```

```python
feature_data
```

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score

# Separate features (X) and target (y)
X = feature_data.drop('label', axis=1)
y = feature_data['label']

# Encode categorical features
for column in X.select_dtypes(include='object').columns:
    le = LabelEncoder()
    X[column] = le.fit_transform(X[column])

# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Initialize and train the Decision Tree Classifier
dtc = DecisionTreeClassifier(random_state=42)
dtc.fit(X_train, y_train)

# Make predictions on the test set
y_pred = dtc.predict(X_test)

# Evaluate the model
accuracy = accuracy_score(y_test, y_pred)
print(f"Decision Tree Classifier Accuracy: {accuracy:.4f}")
```

```python
from sklearn.metrics import classification_report

# Make predictions on the test set
y_pred = dtc.predict(X_test)

# Generate a classification report
report = classification_report(y_test, y_pred)
print("Classification Report:")
print(report)
```

```python
from sklearn.metrics import roc_auc_score

# Get predicted probabilities for the positive class (class 1)
y_pred_proba = dtc.predict_proba(X_test)[:, 1]

# Calculate ROC-AUC score
roc_auc = roc_auc_score(y_test, y_pred_proba)
print(f"ROC-AUC Score: {roc_auc:.4f}")
```
